## Part C:- Filtering with Condition

```python
import pandas as pd

tennis_data = pd.read_csv(r"C:\Users\HP\OneDrive\Desktop\DEALY_PRATICE\00_code_for_GitHub\Pandas-practice-for-Machine-Learning\02_data_analysis\australian_open_2025.csv")

# Q1: Select matches where both: Player1_Rank < 20 AND Player2_Rank < 30
print("Q1 solution: ", tennis_data[(tennis_data["Player1_Rank"]<20)&(tennis_data["Player2_Rank"]<30)])

# Q2: Select matches where: (Player1_Rank < 20 AND Winrate_AI_Pct > 70) OR (Player2_Rank < 20 AND Winrate_AI_Pct > 70)
print("Q2 solution: ", tennis_data[((tennis_data["Player1_Rank"]<20)&(tennis_data["Winrate_AI_Pct"]>70)|(tennis_data["Player2_Rank"]<20)&(tennis_data["Winrate_AI_Pct"]>70))])

# Q3: Select matches where Player 1 is ranked better than Player 2, the AI win rate is above 65%, and 
# the match was played on Hard or Clay surface.
print("Q3 solution: \n",tennis_data[(tennis_data["Player1_Rank"] < tennis_data
["Player2_Rank"]) &(tennis_data["Winrate_AI_Pct"] > 65) &((tennis_data["Surface"]
 == "Hard") |(tennis_data["Surface"] == "Clay"))])
  

# Q4: Select matches where either (Player1_Rank < 15 AND Player2_Rank < 40 AND Winrate_AI_Pct > 60) 
# OR (Player2_Rank < 15 AND Player1_Rank < 40 AND Winrate_AI_Pct > 70), and the Surface is NOT "Grass"..

print("Q4 solution: ", tennis_data[(((tennis_data["Player1_Rank"] < 15) & (tennis_data["Player2_Rank"]
 < 40) & (tennis_data["Winrate_AI_Pct"] > 60)) | ((tennis_data["Player2_Rank"] < 15) & (tennis_data
["Player1_Rank"] < 40) & (tennis_data["Winrate_AI_Pct"] > 70))) & (tennis_data["Surface"] != "Grass")])

# Q5: Select matches where either (Player1_Rank < 10 AND Player2_Rank > 20 AND Winrate_AI_Pct >= 70) 
# OR (Player2_Rank < 10 AND Player1_Rank > 20 AND Winrate_AI_Pct >= 70), and the Surface is either 
# "Hard" OR "Clay".
print("Q5 solution: ", tennis_data[(((tennis_data["Player1_Rank"] < 10) & (tennis_data["Player2_Rank"]
     > 20) & (tennis_data["Winrate_AI_Pct"] >= 70)) | ((tennis_data["Player2_Rank"] < 10) & 
    (tennis_data["Player1_Rank"] > 20) & (tennis_data["Winrate_AI_Pct"] >= 70))) & ((tennis_data
["Surface"] == "Hard") | (tennis_data["Surface"] == "Clay"))])
```
