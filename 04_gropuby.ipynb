## Part D:- Groupby

```python
import pandas as pd


tennis_data = pd.read_csv(r"C:\Users\HP\OneDrive\Desktop\DEALY_PRATICE\00_code_for_GitHub\Pandas-practice-for-Machine-Learning\02_data_analysis\australian_open_2025.csv")

# Q1: Find the average Prize_Money_USD for each Tour.
print("Q1 solution:\n", tennis_data.groupby("Tour")["Prize_Money_USD"].mean())

# Q2: For each Surface, calculate the mean, minimum, and maximum Prize_Money_USD
print("Q2 solution:\n", tennis_data.groupby("Surface")["Prize_Money_USD"].agg(["max","min","mean"]))

# Q3: For each Surface, calculate the average, std and var Sets_Won_P1 and Sets_Won_P2.
print("Q3 solution:\n", tennis_data.groupby("Surface")[["Sets_Won_P1","Sets_Won_P2"]].
      agg(["mean","std","var"]))

# Q4: Group by Tour and Surface. Calculate the average Prize_Money_USD, average Winrate_AI_Pct, and 
# average Match_Reliability for each combination.
print("Q4 solution:\n",  tennis_data.groupby(["Tour","Surface"])[["Prize_Money_USD","Winrate_AI_Pct",
"Match_Reliability"]].mean())

# Q5: For each Surface, calculate the average Sets_Won_P1 and Sets_Won_P2, then calculate the 
# difference between them.
print("Q5 solution:\n", tennis_data.groupby("Surface")[["Sets_Won_P1","Sets_Won_P2"]]
.mean().assign(Sets_Won_Diff=lambda x: x["Sets_Won_P1"] - x["Sets_Won_P2"]))



```
