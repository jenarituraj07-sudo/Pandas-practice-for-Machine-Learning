## Part C:- Boolean Filtering

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Neha", "Ravi", "Sneha", "Arjun", "Pooja"],
    "Age": [21, 24, 22, 27, 25, 23, 29, 26],
    "Salary": [25000, 45000, 32000, 60000, 50000, 38000, 70000, 42000],
    "Experience": [1, 4, 2, 7, 5, 3, 8, 4],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR", "Finance", "IT"],
    "Performance": [78, 88, 72, 95, 91, 80, 97, 86]
})

# Q1: Using Boolean filtering, select employees who have: Salary> 40,000, Experience >= 4, Department is IT
print("Q1 solution:\n", df[(df["Salary"] > 40000)&(df["Experience"] >= 4)&(df["Department"] == "IT")])

# Q2: Select employee who belong to either: "Finance" department OR "HR" department
print("Q2 solution:\n", df[(df["Department"] == "Finance")|(df["Department"] == "HR")])

# Q3: Select employees whose: Age >= 23 AND Age<= 27
print("Q3 solution:\n", df[(df["Age"]>= 23)&(df["Age"]<=27)])

# Q4: Select all employees whose department is not "IT".
print("Q4 solution:\n", df[~(df["Department"] == "IT")])

# Q5: Select employees who satisfy: (Salary > 50,000 AND Performance >= 90) OR (Experience >= 7)
print("Q5 solution:\n", df[((df["Salary"]>50000)&(df["Performance"] >= 90))|(df["Experience"] >= 7)])

# Q6: Select employees who: Department == "IT" AND Salary is between 30,000 and 55,000
print("Q6 solution:\n", df[(df["Department"] == "IT")&(df["Salary"] >= 30000)&(df["Salary"] <= 55000)])

# Q7: Select employees who: Department != "HR" AND Salary > 40,000
print("Q7 solution:\n", df[(df["Department"] != "HR")&(df["Salary"]>40000)])

# Q8: Select employees who satisfy: ( Department == "IT" AND Salary > 40,000 AND Performance >= 85)
#  OR(Department == "Finance" AND Experience >= 7)

print("Q8 solution:\n", df[((df["Department"] == "IT") & (df["Salary"] > 40000) & (df["Performance"]>= 85)) | ((df["Department"] == "Finance") & (df["Experience"] >= 7))][["Name", "Salary", "Experience", "Department", "Performance"]])
```
