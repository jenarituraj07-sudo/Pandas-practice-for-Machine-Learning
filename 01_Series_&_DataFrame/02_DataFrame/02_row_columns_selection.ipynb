## Part B:- Selection for DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Neha", "Ravi", "Sneha", "Arjun", "Pooja"],
    "Age": [21, 24, 22, 27, 25, 23, 29, 26],
    "Salary": [25000, 45000, 32000, 60000, 50000, 38000, 70000, 42000],
    "Experience": [1, 4, 2, 7, 5, 3, 8, 4],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR", "Finance", "IT"],
    "Performance": [78, 88, 72, 95, 91, 80, 97, 86]
})

# Q1: Select Salary Experience Performance with same order.
print("Q1 solution: ", df[["Salary", "Experience", "Performance"]])

# Q2: Select these columns exactly in this order:- Performance Name Salary Department using iloc
print("Q2 solution: ", df.iloc[:,[5,0,2,4]])
print("Q2 solution: ", df.loc[:,["Performance","Name", "Salary", "Department"]])

# Q3: select row:- 1,3,5,7 columns:- Name, Salary, Performance
print("Q3 solution: ", df.iloc[[1,3,5,7],[0,2,-1]])
print("Q4 solution: ", df.loc[[1,3,5,7],["Name", "Salary", "Performance"]])

# Q4: select row:- 0,2,4,6 and columns at position 0,2,4,5 using iloc
print("Q4 solution: ", df.iloc[[0,2,4,6],[0,2,4,5]])
print("Q4 solution: ", df.loc[[0,2,4,6],["Name", "Salary", "Department", "Performance"]])

# Q5: Select employees whose row positions are: 1, 2, 5, 6 and with Name, Age display only 
print("Q5 solution: ", df.iloc[[1,2,5,6],[0,1]])
print("Q5 solution: ", df.loc[[1,2,5,6],["Name","Age"]])

# Q6: Select all employees whose Salary is greater than 45,000.
print("Q6 solution: ",df.loc[df["Salary"] > 45000, ["Name", "Salary", "Department"]])
print("Q6 solution: ",df.iloc[(df["Salary"] > 45000).values, [0, 2, 4]])
```
