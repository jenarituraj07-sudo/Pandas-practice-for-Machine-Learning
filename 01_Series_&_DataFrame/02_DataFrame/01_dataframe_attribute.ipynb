## Part A:- DataFrame Attribute

```python
import pandas as pd

# Q1: Complete Attribute Analysis with function

def dataframe_attributes(df):

    return{
        "shape": df.shape,    #|Return type:- Tuple
        "size": df.size,      #|Return type:- int
        "diesmation": df.ndim,#|Return type:- int
        "index": df.index,    #|Return type:- int (depends on index type)
        "columns": df.columns, #| Return type:- str
        "data_type": df.dtypes,#|Return type:- int64 (depends on index type)

    }

df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Neha", "Ravi"],
    "Age": [21, 24, 22, 27, 25],
    "Salary": [25000, 45000, 32000, 60000, 50000],
    "Experience": [1, 4, 2, 7, 5]
})

df_analysis = dataframe_attributes(df)
# print(df_analysis)
```
