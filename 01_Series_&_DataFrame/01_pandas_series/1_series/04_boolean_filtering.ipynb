## Part D: Boolean Filtering

```python
import pandas as pd

age = pd.Series(
    [18, 25, 32, 41, 29, 55, 67, 23],
    index=["Amit", "Priya", "Rahul", "Neha", "Ravi", "Sneha", "Arjun", "Pooja"],
    name="Age"
)

# Q1: Select customers whose age is greater than 20 and less than 60 and not equal to 25
print("Q1 solution: ", age[(age>20)&(age<60)&(age!=25)])

# Q2: Select customers whose age is less than 25 and 60 greater than 40
print("Q2 solution: ", age[(age<25)|(age>40)&(age<60)])

# Q3: First calculate the median age. Then select all customers whose age is greater than the median age.
median_value = age.median()
print("median value:",median_value)
print("Q3 solution: ", age[age>median_value])

# Q4: greater than the average age AND less than 60
ave_age = age.mean()
print(ave_age)
print("Q4 solution: ", age[(age>ave_age)&(age<60)])

# Q5: Group 1: age >= 18 AND age <= 25 or Group 2: age >= 55 AND age <= 70
print("Q5 solution: ", age[((age >= 18) & (age <= 25))|((age >= 55) & (age <= 70))])
```
