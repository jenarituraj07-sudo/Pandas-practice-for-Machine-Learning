## Part B: Selection Index and Label wise

```python
import pandas as pd

scores = pd.Series(
    [72, 85, 91, 64, 88, 76, 95],
    index=["A", "B", "C", "D", "E", "F", "G"],
    name="Scores"
)

# Q1: Select the score of student C 

print("Q1 solution label wise: ", scores.loc["C"])
print("Q1 solution integer position wise: ", scores.iloc[2])

# Q2: select students A, D, G 
print("Q2 solution label wise: \n", scores.loc[["A","D","G"]])
print("Q2 solution integer position wise: \n", scores.iloc[[0,3,6]])

# Q3: select student B, E, F
print("Q3 solution label wise: \n", scores.loc[["B","E","F"]])
print("Q3 solution integer position wise: \n", scores.iloc[[1,4,5]])

# Q4: Select students C, D, G using:
print("Q4 solution label wise: \n", scores.loc[["C","D","G"]])
print("Q4 solution integer position wise: \n", scores.iloc[[2,3,-1]])

# Q5: Select students A, C, E, G using:
print("Q5 solution label wise: \n", scores.loc[["A","C","E","G"]])
print("Q5 solution integer position wise: \n", scores.iloc[[0,2,4,-1]])


```
