## Part A: Series Creation & Attributes

```python
import pandas as pd

def series_analysis(marks):

    return{
        "shape": marks.shape,       #|return type:- Tuple
        "size": marks.size,         #|return type:- int
        "diesmation": marks.ndim,   #|return type: int
        "data_type": marks.dtype,   #|return type:- int
        "index": marks.index,       #|return type:- int
        "value": marks.values,      #|return type:- int
        "name": marks.name          #|return type:- str
    }

marks = pd.Series(
    [56, 78, 91, 45, 83, 69, 95],
    index=["A", "B", "C", "D", "E", "F", "G"],
    name="Marks"
)
series_stats = series_analysis(marks)
print(series_stats)
```
