## Part C: Series Statistics

```python
import pandas as pd

# Q1: 
salary = pd.Series(
    [25000, 35000, 42000, 28000, 55000, 31000, 48000],
    index=["A", "B", "C", "D", "E", "F", "G"],
    name="Salary"
)

def stats_analysis(salary):

    return{
        "total_salary": salary.sum(),
        "average_salary": salary.mean(),
        "median_salary": salary.median(),
        "minimum_salary": salary.min(),
        "maximun_salary": salary.max(),
        "high_salary_person": salary.idxmax(),
        "high_salary_person": salary.idxmin()
    }

series_analysis = stats_analysis(salary)
print(series_analysis)
```
