## Part B:- Row and Columns Selection

```python
import pandas as pd

tennis_data = pd.read_csv(r"C:\Users\HP\OneDrive\Desktop\DEALY_PRATICE\00_code_for_GitHub\Pandas-practice-for-Machine-Learning\02_exploratory_data_analysis\australian_open_2025.csv")

# Q1: Select the 1st row of the DataFrame using iloc.
print("Q1 solution: \n", tennis_data.iloc[0])

# Q2: Select all rows where Player2_Rank is less than 20. Display Match_ID, Player2_Surname, and 
# Player2_Rank
print("Q2 solution: \n", tennis_data.loc[tennis_data["Player2_Rank"]<20, 
        ["Match_ID","Player2_Surname","Player2_Rank"]])

# Q3: Select all matches where Surface is "Clay". Display Match_ID, Tournament, Surface, and 
# Player1_Surname
print("Q3 solution: \n", tennis_data.loc[tennis_data["Surface"]=="Clay",
                ["Match_ID","Tournament","Surface","Player1_Surname"]])

# Q4: Select all matches where Player1_Rank is less than 10 and Player2_Rank is less than 10. Display 
# both players' surnames and ranks
print("Q4 solution: \n", tennis_data.loc[(tennis_data["Player1_Rank"]<10)&
    (tennis_data["Player2_Rank"]<10),["Player1_Surname", "Player2_Surname", 
    "Player1_Rank", "Player2_Rank"]])

# Q5: Select all matches where Sets_Won_P1 is greater than Sets_Won_P2. Display Match_ID, 
# Player1_Surname, Player2_Surname, Sets_Won_P1, and Sets_Won_P2
print("Q5 solution: \n", tennis_data.loc[tennis_data["Sets_Won_P1"]>
    tennis_data["Sets_Won_P2"],["Match_ID", "Player1_Surname", "Player2_Surname",
     "Sets_Won_P1", "Sets_Won_P2"]])
```
